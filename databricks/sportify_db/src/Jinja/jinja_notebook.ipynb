```python
from jinja2 import Template
```

```python
parameters = [

    {
        "table" : "sportify_cata.silver.factstream",
        "alias" : "factstream",
        "cols" : "factstream.stream_id, factstream.listen_duration"
    },

    {
        "table" : "sportify_cata.silver.dimuser",
        "alias" : "dimuser",
        "cols" : "dimuser.user_id, dimuser.user_name",
        "condition" : "factstream.user_id = dimuser.user_id"
    },
      
    {
        "table" : "sportify_cata.silver.dimtrack",
        "alias" : "dimtrack",
        "cols" : "dimtrack.artist_id, dimtrack.track_name",
        "condition" : "factstream.track_id = dimtrack.track_id"
    
    }
]
```

```python
querry_text = """ 
               
               SELECT 
                   {% for param in parameters %}
                      {{ param.cols }}
                         {% if not loop.last %}
                                 
                                 ,
                         {% endif %}

                   {% endfor %}
              FROM
                {% for param in parameters %}
                   {% if loop.first %}
                      {{param['table'] }}
                   {% endif %}
                {% endfor %}

                {% for param in parameters %}
                   {% if not loop.first %}
                   LEFT JOIN
                      
                         {{param['table'] }} AS {{ param['alias'] }}
                      ON
                         {{ param['condition'] }}
                   {% endif %}

                {% endfor %}

 """ 
```

```python
jinja_sql_str = Template(querry_text)
querry = jinja_sql_str.render(parameters=parameters)
print(querry)
```
